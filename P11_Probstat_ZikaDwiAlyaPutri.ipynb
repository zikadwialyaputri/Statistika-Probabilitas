```python
import re
import pandas as pd
import numpy as np
from scipy import stats
from google.colab import files
```

```python
up = files.upload()
nama_file = list(up.keys())[0]
print("File terbaca:", nama_file)
```

```python
JUMLAH_KOLOM_DEMOGRAFI = 6

df = pd.read_excel(nama_file)
butir = df.iloc[:, JUMLAH_KOLOM_DEMOGRAFI:]
butir = butir.loc[:, butir.columns.str.contains(r"\[", regex=True)]

print("Jumlah responden :", len(df))
print("Jumlah butir     :", butir.shape[1])
print("Data kosong      :", int(butir.isna().sum().sum()))
```

```python
skala = {
    "Sangat Tidak Setuju (STS)": 1,
    "Tidak Setuju (TS)": 2,
    "Netral (N)": 3,
    "Setuju (S)": 4,
    "Sangat Setuju (SS)": 5,
}

num = butir.replace(skala).apply(pd.to_numeric, errors="coerce")

# Harus 0. Jika lebih dari 0, ada jawaban yang tidak terbaca.
print("Sel yang gagal dikonversi:", int(num.isna().sum().sum()))
num.head()
```

```python
dimensi = {}
for kolom in butir.columns:
    nama = re.match(r"^(.*?)\s*\[", kolom).group(1).strip()
    dimensi.setdefault(nama, []).append(kolom)

for nama, daftar in dimensi.items():
    print(f"{nama}: {len(daftar)} butir")
```

```python
n = len(num)
t = stats.t.ppf(0.975, n - 2)
r_tabel = t / np.sqrt((n - 2) + t**2)
print(f"n = {n} | r tabel (α=0,05; df={n-2}) = {r_tabel:.3f}")

hasil_validitas = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    total = x.sum(axis=1)
    for i, kolom in enumerate(daftar, 1):
        r, p = stats.pearsonr(x[kolom], total)
        r_koreksi, _ = stats.pearsonr(x[kolom], total - x[kolom])
        hasil_validitas.append({
            "Dimensi": nama,
            "Butir": f"B{i}",
            "r hitung": round(r, 3),
            "r tabel": round(r_tabel, 3),
            "Corrected r": round(r_koreksi, 3),
            "p-value": round(p, 4),
            "Keterangan": "Valid" if r > r_tabel else "Tidak valid",
        })

tabel_validitas = pd.DataFrame(hasil_validitas)
tabel_validitas
```

```python
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (1 - x.var(ddof=1).sum() / x.sum(axis=1).var(ddof=1))

def kategori(a):
    return ("Sangat baik" if a >= 0.90 else "Baik" if a >= 0.80
            else "Reliabel" if a >= 0.70 else "Tidak reliabel")

hasil_reliabilitas = []
alpha_dihapus = []
for nama, daftar in dimensi.items():
    x = num[daftar]
    a = cronbach_alpha(x)
    hasil_reliabilitas.append({
        "Dimensi": nama,
        "Jumlah butir": len(daftar),
        "Cronbach's alpha": round(a, 3),
        "Kategori": kategori(a),
    })
    for i, kolom in enumerate(daftar, 1):
        if len(daftar) > 2:
            alpha_dihapus.append({
                "Dimensi": nama,
                "Butir": f"B{i}",
                "Alpha jika dihapus": round(cronbach_alpha(x.drop(columns=kolom)), 3),
            })

tabel_reliabilitas = pd.DataFrame(hasil_reliabilitas)
tabel_alpha_dihapus = pd.DataFrame(alpha_dihapus)
display(tabel_reliabilitas)
display(tabel_alpha_dihapus)
```

```python
hasil_deskriptif = []
skor_dimensi = {}

for nama, daftar in dimensi.items():
    skor = num[daftar].mean(axis=1)
    skor_dimensi[nama] = skor
    p_normal = stats.shapiro(skor).pvalue

    hasil_deskriptif.append({
        "Dimensi": nama,
        "Mean": round(skor.mean(), 3),
        "Median": round(skor.median(), 3),
        "Standar Deviasi": round(skor.std(), 3),
        "Varians": round(skor.var(), 3),
        "Skewness": round(skor.skew(), 3),
        "Kurtosis": round(skor.kurt(), 3),
        "p-value Normalitas": round(p_normal, 4),
        "Normalitas": "Tidak normal" if p_normal <= 0.05 else "Tidak menolak normalitas"
    })

tabel_deskriptif = pd.DataFrame(hasil_deskriptif)
display(tabel_deskriptif)
```

```python
import matplotlib.pyplot as plt

for nama, daftar in dimensi.items():
    print("\nDimensi:", nama)

    frekuensi = num[daftar].stack().value_counts().reindex(
        [1, 2, 3, 4, 5], fill_value=0
    )
    display(pd.DataFrame({
        "Jawaban": ["STS", "TS", "Netral", "Setuju", "SS"],
        "Jumlah": frekuensi.values
    }))

    plt.figure(figsize=(7, 4))
    plt.hist(skor_dimensi[nama], bins=8, edgecolor="black")
    plt.axvline(skor_dimensi[nama].mean(), linestyle="--", label="Mean")
    plt.title("Distribusi: " + nama)
    plt.xlabel("Rata-rata Skor")
    plt.ylabel("Jumlah Responden")
    plt.legend()
    plt.show()
```

```python
gabung = tabel_validitas.merge(
    tabel_alpha_dihapus, on=["Dimensi", "Butir"], how="left"
)

with pd.ExcelWriter("hasil_uji_validitas_reliabilitas.xlsx") as w:
    gabung.to_excel(w, sheet_name="Validitas", index=False)
    tabel_reliabilitas.to_excel(w, sheet_name="Reliabilitas", index=False)
    tabel_deskriptif.to_excel(w, sheet_name="Deskriptif", index=False)

files.download("hasil_uji_validitas_reliabilitas.xlsx")
```
